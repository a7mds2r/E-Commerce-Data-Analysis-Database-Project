## dim tables

```python
import pymysql
import csv
import os
```

```python
conn = pymysql.connect(
    host="localhost",                                                                                                      
    user="root",
    password="1998",
    database="E_commerce_db",
    autocommit=True
)
cursor = conn.cursor()
```

```python
file_path = os.path.join("..", "data", "clean_data", "users.csv")
with open(file_path, "r") as f:
    reader = csv.reader(f)
    row = list(reader)

column = row[0]
data_row = row[1:]
cols = ", ".join(column)
placeholders = ", ".join(["%s"] * len(column))

query = f"insert into dim_users ({cols}) values ({placeholders})"

cursor.executemany(query, data_row)
print("successfully: ", cursor.rowcount)
```

```python
cursor.execute("set foreign_key_checks = 0")
cursor.execute("truncate table fact_reviews")
cursor.execute("set foreign_key_checks = 1")
```

```python
file_path = os.path.join("..", "data", "clean_data", "reviews.csv")
with open(file_path, "r") as f:
    reader = csv.reader(f)
    row = list(reader)

column = row[0]
data_row = row[1:]
cols = ", ".join(column)
placeholders = ", ".join(["%s"] * len(column))

query = f"insert into fact_reviews ({cols}) values ({placeholders})"    
try:
    cursor.execute("set foreign_key_checks = 0")

    cursor.executemany(query, data_row)
    print("successfully: ", cursor.rowcount)

    cursor.execute("set foreign_key_checks = 1")
except Exception as e:
    cursor.execute("set foreign_key_checks = 1")
    print(f"error: {e}")
```
